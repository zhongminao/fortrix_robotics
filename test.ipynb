```python
import subprocess

def run_command_safe(command_args):
    """
    安全地执行命令，不经过shell。
    :param command_args: 命令和参数的列表，例如 ["ls", "-l"]
    :return: 命令的输出结果
    """
    try:
        # check=True 会在命令执行失败时抛出异常[reference:6]
        # capture_output=True 用于捕获标准输出和错误[reference:7]
        # text=True 让输出以字符串而非字节形式返回[reference:8]
        # timeout=30 设置超时时间，防止进程卡死[reference:9]
        result = subprocess.run(
            command_args,
            check=True,
            capture_output=True,
            text=True,
            timeout=30
        )
        return result.stdout
    except subprocess.CalledProcessError as e:
        # 命令执行失败（返回码非0）
        return f"命令执行失败，错误码: {e.returncode}, 错误信息: {e.stderr}"
    except subprocess.TimeoutExpired:
        return "命令执行超时"
    except Exception as e:
        return f"发生未知错误: {e}"

# 示例：执行 "ls -l"
output = run_command_safe(["ls", "-l"])
print(output)
```
